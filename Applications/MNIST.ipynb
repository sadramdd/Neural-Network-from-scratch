```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from tensorflow.keras.datasets import mnist
```

```python
(X_train, y_train), (X_test, y_test) = mnist.load_data()
```

```python
print(type(X_train))
print(type(y_train))
```

```python
#Normalize
X_train = X_train / 255.0
X_test = X_test / 255.0
```

```python
#Import model
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent.parent
sys.path.insert(0, str(PROJECT_ROOT))

from deep_neural_network.core.main_structure import NeuralNetwork
```

```python
print(X_train.shape)
print(y_train.shape)

print(X_test.shape)
print(y_test.shape)
```

```python
X_train = X_train.reshape(60000, 784).T
X_test  = X_test.reshape(10000, 784).T

num_classes = 10

y_train = np.eye(num_classes)[y_train].T
y_test = np.eye(num_classes)[y_test].T

print(X_train.shape)
print(y_train.shape)

print(X_test.shape)
print(y_test.shape)
```

```python
net = NeuralNetwork(n_features=X_train.shape[0],
                   task="MultiClassifier",
                   learning_rate= 0.001,
                   random_state= 432)
```

```python
net.add_layer(64, "ReLU", "He", True)
net.add_layer(32, "ReLU", "He", True)
net.add_layer(16, "ReLU", "He", True)
net.add_layer(10, "Softmax", "Xavier", False)
```

```python
net.set_optimizer("Adam")
```

```python
net.fit(X_train,
       y_train,
       200,
       256,
       1)
```

```python
net.evaluate(X_test,
            y_test)
```
